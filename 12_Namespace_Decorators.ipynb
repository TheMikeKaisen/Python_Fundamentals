### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

```python
# local and global
# global var
a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> editing global
a = 2

def temp():
    # Important- Even though you can access a global variable inside local scope
    # But you cannot modify it in like this.
  # local var
  a += 1
  print(a)

temp()
print(a)
```

```python
a = 2

def temp():
  # local var
    # To change a global variable inside a local scope, you have to use global keyword
  global a
  a += 1
  print(a)

temp()
print(a)
```

```python
# local and global -> global created inside local
def temp():
  # local scope
  global a # creating a global variable inside local scope
  a = 1
  print(a)

temp()
print(a)
```

```python
# local and global -> function parameter is local
def temp(z): # here z is a local variable
  # local var
  print(z)

a = 5
temp(5)
print(a)
print(z)
```

```python
# ENCLOSED SCOPE

a = 1 # this variable is in global scope
def outer():
    a = 3 # This variable is in outer() ENCLOSED SCOPE
    def inner():
      a = 6 # this variable is in local scope
      print(a)
    inner()
    print('outer function')


outer()
print('main program')
```

```python
# nonlocal keyword
def outer():
  a = 1
  def inner():
    # a += 2 #ERROR: you cannot change enclosed scope variable inside local scope without nonlocal keyword
    nonlocal a
    a += 1
    print('inner',a)
  inner()
  print('outer',a)


outer()
print('main program')
```

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

Functions are treated just like any other object (e.g., integers, strings, lists).

So in Python, a function can:

✅ Be assigned to a variable

✅ Be passed as an argument to another function

✅ Be returned from another function

✅ Be stored in data structures (lists, dicts, etc.)

```python
# Python are 1st class function

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello nitish')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python
# How does decorators work: CLOSURES IN PYTHON

def outer():
    a = 5
    def inner():
        print(a)
    return inner

b = outer() # outer() returned inner(), thus outer() has done its job
b() # 5 is in outer() scope, outer() has already returned inner(), but still, b() is able to access the value of a
# This is called closure
```

```python
# Better syntax?
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

@my_decorator
def hello():
  print('hello')

hello()
```

```python
# Meaningful Decorator in Python
# creating a timer to find the time taken to execute a function
import time

def timer(func):
  def wrapper(*args): # *args will handle any number of parameters recieved!
    start = time.time()
    func(*args)
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(a,b):
  print(a**b)

hello()
square(2)
power(2,3)
```

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
```
